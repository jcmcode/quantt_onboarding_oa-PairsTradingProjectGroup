```python
df['2xRoom'] = df['Room number'] * 2
df
```

```python
mask = df['Room number'].isna()
df['Text Value'] = np.where(mask, df['Date'], np.nan)
df
```

```python
df['Text Value'].fillna(method='bfill', inplace=True)
df
```

```python
df.dropna(inplace=True)
df
```
